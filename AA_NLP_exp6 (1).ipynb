```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import UnigramTagger, BigramTagger, pos_tag
from nltk.corpus import treebank
from nltk import pos_tag_sents
```

```python
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('punkt_tab')
nltk.download('treebank')
nltk.download('tagsets_json')
```

```python
sample_text = "Apple is planning to open a new store in Mumbai next month."

tokens = word_tokenize(sample_text)
```

```python
rule_based_tags = pos_tag(tokens)
print(rule_based_tags)
```

```python
nltk.help.upenn_tagset("DT")
```

```python
train_sents = treebank.tagged_sents()[:3000]

unigram_tagger = UnigramTagger(train_sents)
unigram_tags = unigram_tagger.tag(tokens)

bigram_tagger = BigramTagger(train_sents, backoff=unigram_tagger)
bigram_tags = bigram_tagger.tag(tokens)
```

```python
print("original sentence :", sample_text)
print("rule based tags :", rule_based_tags)
print("unigram tags :", unigram_tags)
print("bigram tags :", bigram_tags)
```

```python
print("original_text:")
print(sample_text)
print("\nrule_based_tags:")
print(rule_based_tags)
print("\nunigram_tags:")
print(unigram_tags)
print("\nbigram_tags:")
print(bigram_tags)
```

Accuracy=Number of correctly tagged words / Total number of words

```python
test_sents = treebank.tagged_sents()[3000:]
unigram_accuracy = unigram_tagger.evaluate(test_sents)
bigram_accuracy = bigram_tagger.evaluate(test_sents)

print(f"\n unigram accuracy : {unigram_accuracy* 100:.2f}%")
print(f"\n bigram accuracy : {bigram_accuracy*100:.2f}%")
```
