```python
import string
print(string.punctuation)
print("Aaradhya Tandel ,"" UIN :- 231A053")
```

```python
text_lower=text.lower()
print(text_lower)
print("Aaradhya Tandel ,"" UIN :- 231A053")
```

```python
tokens = word_tokenize(text_lower)
print(tokens)
print("Aaradhya Tandel ,"" UIN :- 231A053")
```

```python
text_clean=[]
for word in tokens:
    if word not in string.punctuation:
        text_clean.append(word)
print(text_clean)
print("Aaradhya Tandel ,"" UIN :- 231A053")
```

```python
lemmatizer = WordNetLemmatizer()
lemma = [lemmatizer.lemmatize(word, pos='v') for word in text_clean]
print(lemma)
print("Aaradhya Tandel ,"" UIN :- 231A053")
```

```python
[(tokens,lemma) for tokens,lemma in zip (tokens,lemma)]
print("Aaradhya Tandel ,"" UIN :- 231A053")
```

```python
morphological_info= {}
for tokens,lemma in zip (tokens,lemma):
  if tokens != lemma:
    if tokens.endswith("ing"):
      analysis="Inflactional : present participle/continues form of '{}'".format(lemma)
    elif tokens.endswith("er"):
      analysis="Deriviationsl : '{}'+ -er -> person who does the action".format(lemma)
    else:
      analysis="Inflactional or irregular for of '{}'".format(lemma)
  else:
    analysis="Base/Root form"
morphological_info[tokens]=analysis
print("Aaradhya Tandel ,"" UIN :- 231A053")
```
