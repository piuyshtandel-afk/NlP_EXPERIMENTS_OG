```python
import emoji
text_emoji = emoji.replace_emoji(text,replace="")
print("Aaradhya Tandel","UIN:231A053")
print(text_emoji)
```

Tokenization

```python
import nltk
nltk.download('punkt_tab')
from nltk import word_tokenize
```

```python
tokenized=word_tokenize(text_emoji)
print("Tokenized words",tokenized)
```

```python
print("Aaradhya Tandel","UIN:231A053")
import string
punctuation=list(string.punctuation)
print(punctuation)
```

```python
print("Aaradhya Tandel","UIN:231A053")
text_punct=[]
for word in tokenized:
  if word not in punctuation:
    text_punct.append(word)
print("Without punctuation:",text_punct)
```

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
print((stopwords.words('english')))
```

```python
text_clean=[]
for word in text_punct:
  if word not in stopwords.words('english'):
    text_clean.append(word)
print("Aaradhya Tandel","UIN:231A053")
print("Cleaned text:",text_clean)
```

```python
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
text_stemmed=[stemmer.stem(word) for word in text_clean]
print("Aaradhya Tandel","UIN:231A053")
print("Stemmed text:",text_stemmed)
```

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
lemma=WordNetLemmatizer()
text_lemmatized=[lemma.lemmatize(word) for word in text_clean]
print("Aaradhya Tandel","UIN:231A053")
print("Lemmatized text:",text_lemmatized)
```
